# Regresión con Bosques Aleatorios

## Importar bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importar el dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

## Ajustar el modelo de Regresión Random Forest a TODO el dataset

```python
from sklearn.ensemble import RandomForestRegressor # ensemble: en conjunto
regression = RandomForestRegressor(n_estimators = 100, random_state = 0,criterion='squared_error') 
#n_estimators por defecto =100
#criterion por defecto "mae" pero lo cambio para tener mismo resultado que profesor
regression.fit(X, y)
```

## Predecir un nuevo resultado

```python
regression.predict([[6.5]])
```

## Visualizar los resultados de la Regresión con Random Forest (Mejor resolución)

```python
X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, regression.predict(X_grid), color = 'blue')
plt.title('Regresión con Random Forest')
plt.xlabel('Posición empleado')
plt.ylabel('Salario en $')
plt.show()
```
