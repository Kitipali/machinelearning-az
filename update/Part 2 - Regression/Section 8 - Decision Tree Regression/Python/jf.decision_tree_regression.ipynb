# Regresión de Árboles de Decisión

## Importar bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importar el dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

## Ajustar la regresión con TODO el dataset con Árboles de Decisión

```python
from sklearn.tree import DecisionTreeRegressor
regression = DecisionTreeRegressor(random_state = 0) #random_state = 0 para que coincida el resultado con el del profesor
regression.fit(X, y)
```

## Predicting a new result

```python
regression.predict([[6.5]]) 
# Usamos dobles corchetes porque en scikit-learn, el método .predict() siempre espera una matriz 2D (de tipo array o DataFrame) con forma:
# (n_muestras, n_características)
# Cada fila representa una observación
# Cada columna representa un caractetística (Variable independiente)
```

## Visualización de los resultados del modelo de Árboles de Regresión (alta resolución)

```python
X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, regression.predict(X_grid), color = 'blue')
plt.title('Modelo de Regresión (Árboles de Decisión)')
plt.xlabel('Posición del empleado')
plt.ylabel('Salario en $')
plt.show()
```

Si lo hacemos con todos los datos. Problema de overfitting

```python

X = X.reshape((len(X), 1))
plt.scatter(X, y, color = 'red')
plt.plot(X, regression.predict(X), color = 'blue')
plt.title('Modelo de Regresión (Árboles de Decisión)')
plt.xlabel('Posición del empleado')
plt.ylabel('Salario en $')
plt.show()
```
