# Regresión Logística

## Importar bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importar el dataset

```python
dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

## Dividir el dataset en conjunto de entrenamiento y conjunto de testing

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
print(X_train)
```

```python
print(y_train)
```

```python
print(X_test)
```

```python
print(y_test)
```

## Escalado de variables

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

# No hace falta escalar y porque ya está en formato de 0 y 1
```

```python
print(X_train)
```

```python
print(X_test)
```

## Ajustar el modelo de Regresión Logística en el conjunto de Entrenamiento

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
classifier.fit(X_train, y_train)
```

## Predicting a new result

```python
print(classifier.predict(sc.transform([[30,87000]])))
```

## Predicción de los resultados con el conjunto de Testing

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

## Elaborar la matriz de confusión

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred) # 1º la y verdadera 2º la y predicha
print(cm)
accuracy_score(y_test, y_pred)
```

## Representación gráfica de los resultado del algoritmo de Regresión Logística

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 0.25),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 0.25))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], color = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Clasificador(Conjunto de entrenamiento)')
plt.xlabel('Edad')
plt.ylabel('Salario estimado')
plt.legend()
plt.show()
```

## Visualising the Test set results

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_test), y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 0.25),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 0.25))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Logistic Regression (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```

```python
# Más sencillo con chatgpt

import numpy as np
import matplotlib.pyplot as plt

def plot_decision(clf, X_train, y_train, sc=None,
                  xlabel='Edad', ylabel='Salario estimado', title='Frontera de decisión'):
    # 1) Datos en unidades originales para el gráfico
    Xp = sc.inverse_transform(X_train) if sc is not None else X_train #Si existe un sc (es decir, si no es None), aplica inverse_transform a X_train;
         #si no, usa X_train tal cual.”

    # 2) Malla (rango con pequeño margen)
    x1 = np.linspace(Xp[:,0].min()-2,    Xp[:,0].max()+2,    400)
    x2 = np.linspace(Xp[:,1].min()-1000, Xp[:,1].max()+1000, 400)
    X1, X2 = np.meshgrid(x1, x2)

    # 3) Predicciones en la malla (re-escalando si procede)
    grid = np.concatenate[X1.ravel(), X2.ravel()] #.ravel para que sea 1D
    grid_model = sc.transform(grid) if sc is not None else grid #porque grid está en valores reales pero el clasificador fue entrenado con datos escalados
    Z = clf.predict(grid_model).reshape(X1.shape) #no hay que hacer inverse_transform porque ya lo tenía en cero y uno (la y)

    # 4) Pintar: fondo + puntos
    plt.figure(figsize=(6,4))
    plt.contourf(X1, X2, Z, levels=[-0.5, 0.5, 1.5], alpha=0.6) #levels indica los valores en los que cambian los colores.De -0.5 a 0.5 → zona para clase 0. De 0.5 a 1.5 → zona para clase 1
    # contourf (“contour filled”) rellena con color regiones de un plano según los valores de Z.
    plt.scatter(Xp[:,0], Xp[:,1], c=y_train, s=22, edgecolors='k', linewidths=0.3) #con c=y_train matplotlib determina por defecto que rojo para 0 y verde para 1
    plt.xlabel(xlabel); plt.ylabel(ylabel); plt.title(title); plt.tight_layout(); plt.show()
```

```python
plot_decision(classifier, X_train, y_train, sc=sc,
              xlabel='Edad', ylabel='Salario estimado',
              title='Clasificador (Entrenamiento)')
```
